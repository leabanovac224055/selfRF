```python
import optuna
import optuna.visualization as vis
from optuna.trial import TrialState

# Load the study
study = optuna.load_study(
    study_name="moco_tuning",
    storage="sqlite:////home/airbus/selfRF/moco_tuning.db"
)

# Print all trials and their status
print("\n📋 Trial Summary:")
for t in study.trials:
    print(f"Trial {t.number} — state: {t.state}, value: {t.value}")

# Filter completed trials
completed_trials = [t for t in study.trials if t.state == TrialState.COMPLETE]

if len(completed_trials) == 0:
    print("\n⚠️ No completed trials found. Cannot generate visualizations.")
else:
    print(f"\n✅ Found {len(completed_trials)} completed trial(s). Generating visualizations...")

    # Generate and save visualizations
    fig1 = vis.plot_param_importances(study)
    fig1.write_html("param_importance.html")
    print("📊 Saved: param_importance.html")

    fig2 = vis.plot_optimization_history(study)
    fig2.write_html("optimization_history.html")
    print("📊 Saved: optimization_history.html")

    # Optional: Export trials to CSV
    df = study.trials_dataframe()
    df.to_csv("moco_tuning_results.csv", index=False)
    print("📁 Saved: moco_tuning_results.csv")

```
